```python
import zipfile
import os
from pathlib import Path
# testing how to zip 3mf correctly - using dino in input models
zipList = [f'./input_models/dino/{x}' for x in os.listdir("./input_models/dino")]
print(zipList)
myZip = zipfile.ZipFile("./input_models/dino_test.zip", mode='w') # write mode
for item in zipList:
    myZip.write(item)
myZip.close()

# rename as dino_test.3mf and delete .zip
print(f"Casting zip archives to 3MF")
in_mf_path = Path(f'./input_models/dino_test.zip')
in_mf_path.rename(in_mf_path.with_suffix('.3mf'))
```
