```python
from pygcode import Line
import numpy as np

model_name = "pi 3mm side paint"
# read gcode
pos = np.array([0., 0., 20.])
feed = 5000
gcode_points = []
with open(f'input_gcode/{model_name}_deformed_tet.gcode', 'r') as fh:
    for line_text in fh.readlines():
        line = Line(line_text)

        if not line.block.gcodes:
            continue

        for gcode in sorted(line.block.gcodes):
            if gcode.word.letter == 'T':
                gcode_points.append({
                        "position": pos.copy(),
                        "command": gcode.word,
                        "extrusion": None,
                        "inv_time_feed": None,
                        "move_length": 0.0,
                        "unsegmented_move_length": 0.0,
                        "after_retract": False,
                        "feed": feed
                    })
                continue
```
